Question 6. Median value of horsepower

```python
overall_hp_median=df['horsepower'].median()
overall_hp_median
```

```python
most_frequent=df['horsepower'].mode()[0]
most_frequent
```

```python
df['horsepower']=df['horsepower'].fillna(most_frequent)
df.info()
```

```python
updated_median=df['horsepower'].median()
updated_median
```

```python
result = (
    "Yes, it increased" if updated_median > overall_hp_median
    else "Yes, it decreased" if updated_median < overall_hp_median
    else "No"
)
result
```

Q7. Sum of weights

```python
asian_cars = df[df['origin']=='Asia']
asian_cars.shape
```

```python
X=asian_cars[['vehicle_weight','model_year']].head(7).to_numpy()
X
```

```python
import numpy as np
XTX=X.T @ X
XTX_inv= np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
w
```

```python
sum_w= w.sum()
sum_w
```
